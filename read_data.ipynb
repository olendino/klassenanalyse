```python
raw_data = pd.read_csv('/Users/leonardhaas/code/streamlit/data/statistischer-bericht-verdienste-5623601243225 3.csv',sep=';',header=2)
```

```python
search_term = "Berufsgattungen"

start_row = raw_data[raw_data.apply(lambda row: row.astype(str).str.contains(search_term, case=False).any(), axis=1)].index


```

```python
# Extract the section starting from "Berufsgattungen"
relevant_data = raw_data.iloc[start_row[0]:]

# Clean and restructure the data to extract required columns
# Dropping unnecessary columns and rows with excessive NaN values
cleaned_data = relevant_data.dropna(how='all', axis=1).reset_index(drop=True)
cleaned_data.columns = cleaned_data.iloc[0]  # Set the first row as column headers
cleaned_data = cleaned_data[1:].reset_index(drop=True)  # Remove the header row from the data


cleaned_data.columns = ['bezeichnung_kldb','median_brutto','average_brutto']
structured_data = cleaned_data
""""""
# Clean up numeric fields (removing spaces, handling non-numeric entries)
structured_data['median_brutto'] = structured_data['median_brutto'].str.replace(r'[^\d,]', '', regex=True).str.replace(',', '.').astype(float, errors='ignore')
structured_data['average_brutto'] = structured_data['average_brutto'].str.replace(r'[^\d,]', '', regex=True).str.replace(',', '.').astype(float, errors='ignore')

# Preview the final structured data
```

```python

structured_data.dropna(subset=['bezeichnung_kldb'], inplace=True)
structured_data['code_kldb'] =structured_data['bezeichnung_kldb'].str.extract('(\d+)', expand=False)


```

```python
structured_data['median_brutto'] = pd.to_numeric(structured_data['median_brutto'], errors='coerce').fillna(0).astype(int)
```

```python
structured_data.to_csv('verdienst_destatis_kldb_5steller.csv')
```
